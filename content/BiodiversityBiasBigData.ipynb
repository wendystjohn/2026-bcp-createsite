# 🌿 Investigation 3: Measuring Biodiversity

Now that we understand what information is available, we're ready to begin measuring biodiversity.

Let's start with one of ecology's simplest—and most useful—measurements: **species richness**.

Species richness is simply the **number of different species observed** in a particular area.

Notice that species richness does **not** consider how many individuals of each species were observed. A campus with one observation each of 100 species has the same species richness as a campus with 10,000 observations representing those same 100 species.

Today we'll use species richness as our first measure of biodiversity.

Let's begin by asking a simple question:

> **How many different species have been observed across both campuses?**

 ## 💻 Your Turn

Run the code below to calculate the total number of species observed across both campuses.

```python
# STEP 1
# Calculate total number of species

species = df["scientific_name"].nunique()

print()
print("=" * 45)
print("🌿 RESULT")
print("=" * 45)
print(f"🌿 Total species observed: {species}")
print("=" * 45)
print()
```

### 👀 Take a Look

Great!Now let's ask a slightly different question: 

>How many species have been observed at **each campus**?

## 💻 Run This Cell

⬇️ Run the code below before continuing.

```python
# STEP 2
# Count the number of unique species at each campus

campus_richness = df.groupby("id")["scientific_name"].nunique()

print()
print("=" * 45)
print("🌿 RESULT")
print("=" * 45)
print(f"Indian Valley Campus (IVC): {campus_richness['IVC']} species")
print(f"Kentfield Campus:           {campus_richness['Kentfield']} species")
print("=" * 45)
print()
```

# 🌿 Investigation 4: Biodiversity Through Time

## 🌿 Ecologist's Question

A single species-richness value combines observations collected across many years.

What happens if we separate those observations through time?

> **How has observed species richness changed over time at the two campuses?**

To answer that question, we first need to create a **Year** variable from the observation dates.

## 💻 Step 1: Create a Year Variable

⬇️ Run the code below before continuing.

```python
# Convert observation dates into a format Python recognizes as dates

df["observed_on"] = pd.to_datetime(
    df["observed_on"],
    format="%m/%d/%y"
)

# Create a new variable containing only the year

df["Year"] = df["observed_on"].dt.year

# Preview the result

df[["observed_on", "Year"]].head()
```

> 💡 **Computational Note**
>
> We just created a **new variable** from information that was already present in the dataset.
>
> Computational tools often allow scientists to transform existing data into new variables that help answer a particular question.

## 💻 Step 2: Calculate Richness by Year

Now we'll calculate the number of unique species observed at each campus during each year.

⬇️ Run the code below before continuing.

```python
richness = (
    df.groupby(["Year", "id"])["scientific_name"]
      .nunique()
      .reset_index(name="Species_Richness")
)

# Preview the first few results

richness.head(10)
```

### 👀 Take a Look

Each row now represents the observed species richness for **one campus during one year**.

The table is useful, but patterns across many years are much easier to recognize visually.

Let's make a graph.

## 💻 Step 3: Visualize the Pattern

⬇️ Run the code below before continuing.

```python
for campus in richness["id"].unique():
    subset = richness[richness["id"] == campus]

    plt.plot(
        subset["Year"],
        subset["Species_Richness"],
        marker="o",
        label=campus
    )

plt.xlabel("Year")
plt.ylabel("Observed Species Richness")
plt.title("Observed Species Richness Through Time")
plt.legend(title="Campus")
plt.show()
```

# 🌿 Investigation 5: Is There More to the Story?

The graph reveals some striking changes in observed species richness through time.

But before we explain those changes biologically, good scientists ask:

> **Could something about the way the observations were collected also help explain the pattern?**

## 🌿 Ecologist's Question

The graph from Investigation 4 appeared to answer our original question.

But good scientists don't stop after a single analysis.

Instead, they ask:

> **Could another factor help explain the pattern we observed?**

Imagine surveying birds in a forest.

Would you find more species if you searched for:

- 10 minutes?
- 1 hour?
- 10 hours?

The amount of searching that goes into collecting observations is called **sampling effort**.

Let's investigate whether differences in sampling effort might help explain the biodiversity patterns we observed.

## 💻 Step 1: Count the Observations

⬇️ Run the code below before continuing.

```python
observations = (
    df.groupby(["Year", "id"])
      .size()
      .reset_index(name="Observations")
)

observations.head()
```

### 👀 Take a Look

Each row now shows the total number of observations recorded for one campus during one year. The table above just shows the first few rows of the dataset, so let's see what happens when we graph all the observations.

## 💻 Step 2: Graph Observation Effort

⬇️ Run the code below before continuing.

```python
for campus in observations["id"].unique():
    subset = observations[observations["id"] == campus]

    plt.plot(
        subset["Year"],
        subset["Observations"],
        marker="o",
        label=campus
    )

plt.xlabel("Year")
plt.ylabel("Number of Observations")
plt.title("Observation Effort Through Time")
plt.legend(title="Campus")
plt.show()
```

### 👀 Take a Look

Study the graph carefully.

What happens when we compare observation effort directly with species richness? We'll do this in two steps. 

## 💻 Step 3: Merge the Year and ID fields


⬇️ Run the code below before continuing.

```python
comparison = richness.merge(
    observations,
    on=["Year", "id"]
)

comparison.head()
```

## 💻 Step 4: Compare the Two Measurements

⬇️ Run the code below before continuing.

```python
for campus in comparison["id"].unique():
    subset = comparison[comparison["id"] == campus]

    plt.scatter(
        subset["Observations"],
        subset["Species_Richness"],
        label=campus
    )

plt.xlabel("Number of Observations")
plt.ylabel("Species Richness")
plt.title("Species Richness vs. Observation Effort")
plt.legend(title="Campus")
plt.show()
```

# 🌿 Investigation 6: Looking More Closely

We've discovered that the number of species observed depends not only on the organisms present...

...but also on **how much people looked for them.**

Even so, one question remains.

Suppose two campuses contain similar numbers of species.

>**Would they necessarily contain the same species?**

## 🌿 Ecologist's Question

So far, we've compared the two campuses using **species richness**.

Species richness tells us **how many** species were observed.

But biodiversity isn't only about **how many** species are present.

It's also about **which** species are present.

Imagine two forests that each contain 100 species.

Would they necessarily contain the same 100 species?

Probably not.

Let's compare the communities found at the two campuses.

## 💻 Step 1: Build Species Lists

⬇️ Run the code below before continuing.

```python
# Create a list of species observed at each campus

ivc_species = set(
    df[df["id"] == "IVC"]["scientific_name"]
)

kentfield_species = set(
    df[df["id"] == "Kentfield"]["scientific_name"]
)

print()
print("=" * 45)
print("🌿 RESULT")
print("=" * 45)
print(f"Indian Valley Campus: {len(ivc_species)} species")
print(f"Kentfield Campus:     {len(kentfield_species)} species")
print("=" * 45)
print()
```

### 👀 Take a Look

Although the two campuses have similar species richness...

that doesn't necessarily mean they contain the same species.

Let's compare the two species lists directly.

## 💻 Step 1: Compare Species Lists

⬇️ Run the code below before continuing.

```python
shared = len(ivc_species & kentfield_species)
ivc_only = len(ivc_species - kentfield_species)
kentfield_only = len(kentfield_species - ivc_species)

print()
print("=" * 45)
print("🌿 RESULT")
print("=" * 45)
print(f"Shared species:   {shared}")
print(f"IVC only:         {ivc_only}")
print(f"Kentfield only:   {kentfield_only}")
print("=" * 45)
print()
```

Ecologists have developed a way to summarize this comparison.

One commonly used metric is the **Jaccard Similarity Index**, which is calculated as:

\[
\text{Jaccard Similarity} =
\frac{\text{Number of Shared Species}}
{\text{Total Number of Species}}
\]

A value close to **1** indicates that two communities are very similar.

A value close to **0** indicates that they share relatively few species.

> ### 📘 Concept Check: What Does "Similarity" Mean?
>
> When thinking about how similar or different two different habitats might be, Ecologists often summarize community similarity using an **index**. An index is a numerical measure that summarizes one or more characteristics of a system, making it easier to compare different systems.
>
> In ecology, an index often combines information about a biological community into a single value. This makes it easier to compare communities, even when they contain many different species.
>
> Imagine that we create a similarity index that ranges from **0 to 1**.
>
> - If two communities share **all** of their species, what value should the index have?
> - If two communities share **none** of their species, what value should the index have?
>
> A **Jaccard Similarity Index** works this way. The Jaccard Similarity Index is one example of an ecological index. It summarizes the similarity between two communities using a single number between 0 and 1.
>
> - **1.0** = the communities contain exactly the same species
> - **0.0** = the communities share no species
> - Values in between indicate partial overlap
>
> For example:
>
> **High similarity**
>
> Community A: oak, deer, scrub-jay, salamander  
> Community B: oak, deer, scrub-jay, salamander  
>
> Jaccard = **1.0**
>
> **Intermediate similarity**
>
> Community A: oak, deer, scrub-jay, salamander  
> Community B: oak, deer, egret, frog  
>
> Shared species = 2  
> Total species = 6  
> Jaccard = **2 / 6 = 0.33**
>
> **Low similarity**
>
> Community A: oak, deer, scrub-jay, salamander  
> Community B: cattail, pelican, crab, rail  
>
> Jaccard = **0.0**
>

### 💻 Note About Computational Tools in Biology

Earlier in this investigation, we calculated **species richness** ourselves because it is a relatively simple measurement—the number of unique species observed.

Many ecological analyses, however, involve more complex calculations. Rather than performing these calculations by hand every time, scientists often use **specialized software libraries** that have been developed, tested, and shared by the scientific community.

The **Jaccard Similarity Index** is simple enough that we can calculate it ourselves using Python. Because we understand the formula, we can translate it directly into code.

Notice how the calculation below closely matches the mathematical formula introduced above. As you continue learning Python, you'll discover that many scientific calculations can be expressed just as clearly in code.

Our goal is not simply to obtain a number—it is to understand **what that number means** and determine whether it helps answer our ecological question.

```python
# Calculate the Jaccard Similarity Index
# (Shared species ÷ Total species)

jaccard = shared / total_species

print()
print("=" * 45)
print("🌿 RESULT")
print("=" * 45)
print(f"Jaccard Similarity Index: {jaccard:.3f}")
print("=" * 45)
print()
```

We've learned that the campuses contain surprisingly different biological communities. But biodiversity isn't just a collection of numbers.

## 🌿 Ecologist's Question

The two campuses contain surprisingly different biological communities.

But what kinds of organisms account for those differences?

Let's begin by looking closely at one familiar group of organisms: birds.

## 💻 Step 1: Build Bird Lists

⬇️ Run the code below before continuing.

```python
ivc_birds = set(
    df[
        (df["id"] == "IVC") &
        (df["iconic_taxon_name"] == "Aves")
    ]["common_name"]
)

kentfield_birds = set(
    df[
        (df["id"] == "Kentfield") &
        (df["iconic_taxon_name"] == "Aves")
    ]["common_name"]
)
```

> 💡 **Computational Note**
>
> This code creates two bird species lists—one for each campus.
>
> You won't see any output yet because we're preparing information that we'll compare in the next step.

## 💻 Step 2: Compare the Bird Communities¶

⬇️ Run the code below before continuing.

```python
bird_comparison = pd.DataFrame({
    "Unique to IVC": pd.Series(sorted(ivc_birds - kentfield_birds)),
    "Shared Species": pd.Series(sorted(ivc_birds & kentfield_birds)),
    "Unique to Kentfield": pd.Series(sorted(kentfield_birds - ivc_birds))
})

bird_comparison
```
