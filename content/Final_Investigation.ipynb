## Community Science as Scientific Evidence

The iNaturalist dataset did not make a mistake.

Instead, it reminds us that ecological datasets document **observations**, not every organism that exists within an ecosystem.

Whether a species appears in a dataset depends on many factors, including:

- whether someone observed it,
- whether it could be photographed,
- whether it was uploaded,
- whether it could be identified,
- whether it reached research-grade status.

As scientists, we must always distinguish between:

- **the organisms that are present**, and
- **the organisms represented in our dataset.**
